```python
import pandas as pd
import wget
from zipfile import ZipFile
```

```python
cvm = "https://dados.cvm.gov.br/"
url = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/DFP/DADOS/"
```

```python
arq_zip = []

for ano in range(2010, 2026):
  arq_zip.append(f'dfp_cia_aberta_{ano}.zip')

arq_zip
```

```python
for arq in arq_zip:
  wget.download(url+arq)
```

```python
for arq in arq_zip:
  ZipFile(arq, 'r').extractall('CVM')
```

```python
nomes = [
    # 'BPA_con', 'BPA_ind'
    # ,'BPP_con', 'BPP_ind'
    # ,'DFC_MD_con', 'DFC_MD_ind'
    # ,'DFC_MI_con', 'DFC_MI_ind'
    # ,'DMPL_con', 'DMPL_ind'
    # ,'DRA_con', 'DRA_ind'
    'DRE_con', 'DRE_ind'
    # ,'DVA_con', 'DVA_ind'
    ]

for nome in nomes:
  arquivo = pd.DataFrame()
  for ano in range(2011, 2026):
    arquivo = pd.concat([arquivo, pd.read_csv(f'/content/CVM/dfp_cia_aberta_{nome}_{ano}.csv', sep = ';', decimal = ',', encoding = 'ISO-8859-1')])
  arquivo.to_csv(f'DADOS/dfp_cia_aberta_{nome}_2010-2025', index = False)
```
